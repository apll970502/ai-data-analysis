## 62. 반복 점검 함수

```python
def inspect_dataframe(
    name: str,
    df: pd.DataFrame,
    primary_key: str | None = None,
) -> dict:
    result = {
        "dataset": name,
        "rows": df.shape[0],
        "columns": df.shape[1],
        "column_names": df.columns.tolist(),
        "dtypes": {
            column: str(dtype)
            for column, dtype in df.dtypes.items()
        },
        "missing_total": int(df.isna().sum().sum()),
        "duplicate_rows": int(df.duplicated().sum()),
    }
    if primary_key is not None:
        result["primary_key"] = primary_key
        result["key_missing"] = int(
            df[primary_key].isna().sum()
        )
        result["key_duplicates"] = int(
            df[primary_key].duplicated().sum()
        )
        result["key_unique"] = int(
            df[primary_key].nunique()
        )
    return result
```

```python
inspection_results = []

for name, df in datasets.items():
    inspection_results.append(
        inspect_dataframe(
            name,
            df,
            primary_keys[name],
        )
    )

inspection_results
```

```python
# 63. LLM에 전달할 안전한 구조 요약
safe_schema_summary = {
    "datasets": inspection_results,
    "relationships": {
        "orders.customer_id": {
            "references": "customers.customer_id",
            "unknown_count": len(unknown_customer_ids),
        },
        "order_items.order_id": {
            "references": "orders.order_id",
            "unknown_count": len(unknown_order_ids),
        },
        "order_items.product_id": {
            "references": "products.product_id",
            "unknown_count": len(unknown_product_ids),
        },
    },
}
print(safe_schema_summary)
```

## 1. 데이터셋별 행의 의미

| 데이터셋          | 행의 의미                | 주요 식별자          |
| ------------- | -------------------- | --------------- |
| `customers`   | 고객 1명을 의미            | `customer_id`   |

| `order_items` | 주문에 포함된 상품 항목 1개를 의미 | `order_item_id` |

* `orders`는 **주문 단위**, `order_items`는 **주문에 포함된 상품 항목 단위**로 데이터의 집계 단위가 다름.
* `orders` 301건, `order_items` 764건으로 주문당 평균 약 **2.54개의 상품 항목**이 존재함.
* 따라서 향후 매출 및 판매량 분석 시 어떤 데이터셋을 기준으로 집계하는지 구분할 필요가 있음.

## 2. 기본키 및 외래키 구조

### 기본키(Primary Key)

| 데이터셋          | 기본키             | 점검 결과         |
| ------------- | --------------- | ------------- |
| `customers`   | `customer_id`   | 결측 0건 / 중복 0건 |
| `products`    | `product_id`    | 결측 0건 / 중복 0건 |
| `orders`      | `order_id`      | 결측 0건 / 중복 0건 |
| `order_items` | `order_item_id` | 결측 0건 / 중복 0건 |

각 데이터셋의 기본키는 행 수와 고유값 수가 동일하여 현재 구조상 기본키 문제는 발견되지 않음.

### 외래키(Foreign Key)

| 외래키                      | 참조 대상                   | 참조 오류 |
| ------------------------ | ----------------------- | ----: |
| `orders.customer_id`     | `customers.customer_id` |    0건 |
| `order_items.order_id`   | `orders.order_id`       |    0건 |
| `order_items.product_id` | `products.product_id`   |    0건 |

현재 확인된 범위에서는 외래키의 미등록 참조가 없어 **참조 무결성이 정상적으로 유지되고 있음**.

> `orders.customer_id`는 외래키이므로 여러 번 나타날 수 있으며, 동일 고객의 반복 주문은 정상적인 데이터일 수 있음.

## 3. 결측치 및 중복 데이터 처리 시 주의사항

현재 네 데이터셋 모두 **결측치 0건**, **전체 행 중복 0건**으로 확인됨.

따라서 현재 단계에서는 삭제할 결측치나 중복 데이터가 없음.

향후 결측치나 중복이 발견되더라도 즉시 삭제하지 않고 발생 원인을 먼저 확인해야 함.

### 중복으로 보이지만 정상일 수 있는 사례

* `orders.customer_id`의 반복
  → 동일 고객의 재구매일 가능성이 있음.
* `order_items`에서 동일한 `order_id + product_id`가 반복
  → 실제 주문 구조에 따라 정상적인 기록일 가능성이 있으므로 원인 확인 필요.
* `customers.name`의 반복
  → 동명이인일 가능성이 있으므로 이름만으로 중복 고객으로 판단하면 안 됨.

### 결측치 처리 시 주의

* 결측값이 실제 데이터 부재를 의미하는지 확인할 필요가 있음.
* 특정 `order_status`나 거래 상황 때문에 발생한 결측값일 가능성도 있으므로 원인 확인 후 처리해야 함.

**원칙: 결측치나 중복이 발견되더라도 원인을 확인하기 전에는 임의로 삭제하지 않는다.**

## 6. 종합 판단

현재 구조 점검 결과, **기본키·결측치·전체 행 중복·외래키 참조 무결성에서는 명확한 문제가 발견되지 않음.**

다만 `orders`의 행 수가 이전 점검 결과와 달라졌으므로 **현재 사용 중인 데이터가 최신 데이터인지 먼저 확인할 필요가 있음.**

이후 다음 순서로 추가 점검을 진행하는 것이 적절함.

**① 데이터 기준점 확인 → ② 상품 가격 비교 → ③ 가입일/주문일 비교 → ④ 주문 상태별 금액 확인 → ⑤ 상품 항목이 없는 주문 확인 → ⑥ 주문 상품 조합 중복 확인 → ⑦ 미주문 고객 및 미판매 상품 확인**

또한 현재 확인되지 않은 내용을 데이터 오류로 단정하지 않고, **'현재 확인된 사실'과 '추가 검증이 필요한 항목'을 구분하여 기록하는 것이 중요함.**
